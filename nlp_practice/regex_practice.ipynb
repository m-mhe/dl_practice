### How search works:

```python
# Let's find a person whos last phone number digits are 456
for line in doc:
    match = re.search(r"\d{7,8}456", line)
    if match:
        searchItem = match
searchItem, searchItem.group()
```

### Let's do the same with compile

```python
# Let's find a person whos last phone number digits are 456

searchedNumPatten = re.compile(r"\d{7,8}678")
searchedNamePattern = re.compile(r"name\sis\s(\w+)")

for line in doc:
    match = re.search(searchedNumPatten, line)
    if match:
        name = re.findall(searchedNamePattern, line)
        searchItem = match
name[0], searchItem.group()
```

### How .findall() works

```python
validEmailPattern = re.compile(r"\w+@..+\.\w\w+")
validNumPattern = re.compile(r"\+880\s\d{10}|01\d{9}")
namePattern = re.compile(r"\s([A-Z]\w+)")

candidateInfo = []

for candidate in doc:
    name = re.findall(namePattern, candidate)
    email = re.findall(validEmailPattern, candidate)
    phone = re.findall(validNumPattern, candidate)
    candidateInfo.append({"Name": name, "Email": email, "Phone": phone})

for candidate in candidateInfo:
    for label, info in candidate.items():
        print(f"{label}= {info[0]}")
    print("==========================\n")
```
