# Sgmoid

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
X = np.arange(-10,10, 0.00001)
y = 1 / (1 + np.exp(-X))
dy = y * (1 - y)
X, y, dy
```

```python
plt.figure(figsize=(20,10))
plt.axis([-10, 10, -1.2, 1.2])
plt.plot(X,X, label="f(X)")
plt.plot(X, y, label="σ(x)")
plt.plot(X, dy, label = "σ'(x)")
plt.legend(loc="best")
plt.grid(True)
plt.show()
```
