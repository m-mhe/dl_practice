```python
import torch
```

```python
torch.__version__
```

```python
def bCE(yHat, y):
    e = 1e-8
    yHat = yHat.clamp(e, (1-e))
    loss = -((y * (torch.log(yHat))) + ((1-y) * (torch.log(1-yHat))))
    return loss
```

```python
x = torch.tensor([6.50, 1.0, 5.0, 6.7], dtype=torch.float32)
y = torch.tensor([1, 0, 1, 0], dtype=torch.float32)
type(x), y.shape, x.dtype 
```

```python
w = torch.ones((1), requires_grad=True, dtype=torch.float32)
b = torch.zeros((1), requires_grad=True, dtype=torch.float32)
```

```python
z = w * x + b
z
```

```python
yHat = torch.sigmoid(z)
yHat
```

```python
loss = bCE(yHat, y)
loss
```

```python
loss = loss.mean()
loss
```

```python
loss.backward()
```

```python
w.grad
```

```python
b.grad
```

So as we can see, torch calculates derivative autometically and faster by using the AutoGrad method. Autograd can be thought of as a method/system for automatically computing derivatives, but in PyTorch it's more precise to call it an automatic differentiation engine. 

Autograd is PyTorch’s **automatic differentiation engine**; it automatically calculates derivatives (gradients) of tensors with respect to other tensors, so you don't have to derive them manually. It tracks the mathematical operations performed on tensors that have `requires_grad=True`, builds a **computation graph**, and when you call `loss.backward()`, it traverses that graph backward using the chain rule to calculate the gradients. So, **automatic differentiation is the general technique, while Autograd is PyTorch's implementation of that technique**.
